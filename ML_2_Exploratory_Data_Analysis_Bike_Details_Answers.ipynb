## Question 1. Read the Bike Details dataset into a Pandas DataFrame and display its first 10 rows. (Show the shape and column names as well.)

Answer: Load the provided Bike Details CSV into a DataFrame and inspect the first ten records, dimensions and field names. The actual values depend on the downloaded dataset; no rows or counts are assumed here.

## Question 3. Plot the distribution of selling prices using a histogram and describe the overall trend.

Answer: A selling-price histogram shows how frequently different price ranges occur. Look for a concentration at lower prices, a long right tail and unusually high prices; describe the actual shape after running the plot.

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(data=df, x='selling_price', bins=30, kde=True)
plt.title('Distribution of bike selling prices')
plt.xlabel('Selling price'); plt.tight_layout(); plt.show()
```

## Question 4. Create a bar plot to visualize the average selling price for each seller_type and write one observation.

Answer: Group by seller_type to calculate the mean selling price for each seller category. The highest bar identifies the category with the highest observed average; this is descriptive and does not prove seller type causes the difference.

```python
avg = df.groupby('seller_type')['selling_price'].mean().sort_values()
print(avg)
avg.plot(kind='bar', title='Average price by seller type')
plt.ylabel('Average selling price'); plt.tight_layout(); plt.show()
```

## Question 5. Compute the average km_driven for each ownership type (1st owner, 2nd owner, etc.), and present the result as a bar plot.

Answer: Compute the average kilometers driven for each ownership group and compare the bars. Differences may reflect usage patterns and sample composition rather than ownership alone.

```python
avg_km = df.groupby('owner')['km_driven'].mean().sort_values()
print(avg_km)
avg_km.plot(kind='bar', title='Average km driven by ownership')
plt.ylabel('Average km driven'); plt.tight_layout(); plt.show()
```

## Question 6. Use the IQR method to detect and remove outliers from the km_driven column. Show before-and-after summary statistics.

Answer: The interquartile range (IQR) equals Q3 minus Q1. Flag values outside Q1 − 1.5×IQR and Q3 + 1.5×IQR, compare descriptive statistics, and avoid deleting valid high-mileage bikes without review.

```python
print('Before:\n', df['km_driven'].describe())
q1, q3 = df['km_driven'].quantile([.25, .75])
iqr = q3 - q1
low, high = q1 - 1.5*iqr, q3 + 1.5*iqr
clean = df[df['km_driven'].between(low, high)].copy()
print('After:\n', clean['km_driven'].describe())
print('Removed:', len(df)-len(clean))
```

## Question 7. Create a scatter plot of year vs. selling_price to explore the relationship between a bike's age and its price.​

Answer: Plot manufacturing year against selling price to check whether newer bikes generally sell for more. A visible upward trend would suggest an association, but mileage, model and condition may also influence price.

```python
sns.scatterplot(data=df, x='year', y='selling_price', alpha=.6)
plt.title('Manufacturing year vs selling price')
plt.tight_layout(); plt.show()
```

## Question 8. Convert the seller_type column into numeric format using one-hot encoding. Display the first 5 rows of the resulting DataFrame.

Answer: One-hot encoding converts seller_type into separate indicator columns without imposing an artificial ranking. Display the first five rows to verify the transformation.

```python
encoded = pd.get_dummies(df, columns=['seller_type'], dtype=int)
print(encoded.head().to_string(index=False))
```

## Question 9. Generate a heatmap of the correlation matrix for all numeric columns. What correlations stand out the most?

Answer: A correlation heatmap summarizes pairwise linear relationships between numeric fields. Values near +1 indicate positive association, near −1 negative association, and near 0 little linear association. Report the strongest actual correlations after examining the output.

```python
numeric = df.select_dtypes(include='number')
corr = numeric.corr()
print(corr['selling_price'].sort_values() if 'selling_price' in corr else corr)
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Numeric feature correlations'); plt.tight_layout(); plt.show()
```
