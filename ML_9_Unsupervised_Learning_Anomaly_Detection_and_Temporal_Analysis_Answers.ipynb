## Question 1. What is Dimensionality Reduction? Why is it important in machine learning?

Answer: Dimensionality reduction transforms a dataset with many features into fewer informative variables. It can reduce computation, improve visualization and limit noise or redundancy. However, some information may be lost, so choose the number of dimensions based on explained variance or downstream performance.

## Question 2. Name and briefly describe three common dimensionality reduction techniques..

Answer: Principal Component Analysis (PCA) creates orthogonal directions of maximum variance. t-SNE preserves local similarities for low-dimensional visualization but is generally not a predictive feature transformer. UMAP is another nonlinear method that can preserve local neighborhood structure and often scales well to large datasets.

## Question 3. What is clustering in unsupervised learning? Mention three popular clustering algorithms.

Answer: Clustering groups observations so that members of the same group are more similar than those in different groups, without labeled targets. K-means groups points around centroids; hierarchical clustering builds a nested cluster tree; DBSCAN finds dense regions and can mark sparse observations as noise.

## Question 4. Explain the concept of anomaly detection and its significance.

Answer: Anomaly detection identifies observations that differ substantially from expected behavior. It is useful for fraud alerts, equipment fault detection, cybersecurity monitoring and quality control. Unusual observations are not automatically errors: domain context and false-alarm rates matter.

## Question 5. List and briefly describe three types of anomaly detection techniques.

Answer: Statistical methods flag observations with extreme z-scores or low probability under an assumed distribution. Distance- or density-based methods, such as Local Outlier Factor, identify points far from their neighbors. Model-based methods, such as Isolation Forest, isolate unusual observations using random partitions.

## Question 6. What is time series analysis? Mention two key components of time series data.

Answer: Time series analysis studies observations ordered over time, often to understand patterns and forecast future values. Trend describes the longer-term direction, while seasonality describes regularly repeating patterns such as weekly sales cycles. Noise and cyclical movements may also be present.

## Question 7. Describe the difference between seasonality and cyclic behavior in time series.

Answer: Seasonality repeats at a relatively fixed and known frequency, such as higher weekend traffic each week. Cyclical behavior consists of longer expansions and contractions without a strictly fixed period, such as business cycles. Both create variation over time, but seasonality is generally easier to represent with calendar-based features.

## Question 8. Write Python code to perform K-means clustering on a sample dataset.​

Answer: K-means partitions observations into K groups by repeatedly assigning each point to the nearest centroid and updating centroid locations. The following example generates three synthetic clusters and visualizes the assignments.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
X,_=make_blobs(n_samples=200,centers=3,cluster_std=.7,random_state=42)
model=KMeans(n_clusters=3,random_state=42,n_init=10)
labels=model.fit_predict(X)
print('Cluster counts:',__import__('numpy').bincount(labels))
plt.scatter(X[:,0],X[:,1],c=labels,cmap='viridis',s=20)
plt.scatter(model.cluster_centers_[:,0],model.cluster_centers_[:,1],
            marker='X',s=150,c='red')
plt.title('K-means clustering');plt.show()
```

## Question 9. What is inheritance in OOP? Provide a simple example in Python.

Answer: Inheritance in object-oriented programming allows a child class to reuse and extend behavior from a parent class. It reduces repeated code and supports specialization. In this example, Dog inherits the speak method and overrides it.

```python
class Animal:
    def speak(self):
        return 'Some animal sound'

class Dog(Animal):
    def speak(self):
        return 'Woof!'

pet = Dog()
print(pet.speak())  # Woof!
```

## Question 10. How can time series analysis be used for anomaly detection?

Answer: Time series anomaly detection compares current observations with expected temporal patterns. A simple approach fits a rolling baseline or seasonal forecasting model and flags unusually large residuals. For example, a sudden transaction spike may be suspicious after accounting for normal daily and weekly cycles. Thresholds should be validated against historical alerts to manage false positives.
