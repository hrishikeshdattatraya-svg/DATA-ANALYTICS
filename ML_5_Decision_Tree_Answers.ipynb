## Question 1. What is a Decision Tree, and how does it work in the context of classification?

Answer: A Decision Tree recursively divides observations using feature-based rules. At each internal node, the algorithm selects a split that improves class purity; a leaf predicts a class based on its training examples. For instance, a loan decision tree may first split by credit score and then by income. Trees are interpretable but can overfit when grown too deep.

## Question 2. Explain the concepts of Gini Impurity and Entropy as impurity measures. How do they impact the splits in a Decision Tree?

Answer: Gini impurity is 1 − Σpᵢ², while entropy is −Σpᵢ log₂(pᵢ), where pᵢ is the fraction of class i in a node. Both equal zero for a pure node and increase as classes mix. A decision tree chooses splits that reduce weighted impurity, so the children become more homogeneous.

## Question 3. What is the difference between Pre-Pruning and Post-Pruning in Decision Trees? Give one practical advantage of using each.

Answer: Pre-pruning stops a tree from becoming overly complex during training using limits such as max_depth, min_samples_leaf or min_samples_split. Its advantage is reduced training cost and overfitting. Post-pruning first grows a larger tree and then removes branches that contribute little predictive value, for example using cost-complexity pruning; it can retain useful structure before simplifying.

## Question 4. What is Information Gain in Decision Trees, and why is it important for choosing the best split?

Answer: Information Gain measures how much a split reduces uncertainty: parent entropy minus the weighted average entropy of the child nodes. A higher gain indicates that the split separates class labels more effectively. It guides split selection, although trees may also use Gini impurity instead of entropy.

## Question 5. What are some common real-world applications of Decision Trees, and what are their main advantages and limitations?

Answer: Decision Trees are used in credit-risk screening, customer churn prediction, medical triage support and product recommendation rules. Advantages include clear decision paths, nonlinear splits and little need for feature scaling. Limitations include instability, sensitivity to noise and overfitting; ensembles often improve predictive performance.

## Question 6. Write a Python program to: ●​ Load the Iris Dataset ●​ Train a Decision Tree Classifier using the Gini criterion ●​ Print the model’s accuracy and feature importances

Answer: Fit a Gini-based classifier to Iris and report test accuracy and the relative importance of each feature.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
iris=load_iris(); X,y=iris.data,iris.target
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,stratify=y,random_state=42)
model=DecisionTreeClassifier(criterion='gini',random_state=42)
model.fit(Xtr,ytr)
print('Accuracy:',accuracy_score(yte,model.predict(Xte)))
print(dict(zip(iris.feature_names,model.feature_importances_)))
```

## Question 7. Write a Python program to: ●​ Load the Iris Dataset ●​ Train a Decision Tree Classifier with max_depth=3 and compare its accuracy to a fully-grown tree.

Answer: A depth limit can reduce overfitting. Compare held-out accuracy for max_depth=3 against a fully grown tree using the same split.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
X,y=load_iris(return_X_y=True)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,stratify=y,random_state=42)
for depth in [3,None]:
    model=DecisionTreeClassifier(max_depth=depth,random_state=42).fit(Xtr,ytr)
    print('max_depth:',depth,'test accuracy:',model.score(Xte,yte))
```

## Question 8. Write a Python program to: ●​ Load the California Housing dataset from sklearn ●​ Train a Decision Tree Regressor ●​ Print the Mean Squared Error (MSE) and feature importances

Answer: Use California Housing, the dataset explicitly requested in Question 8, to fit a regression tree. The older Boston Housing dataset mentioned in the assignment is no longer bundled in current scikit-learn versions.

```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
housing=fetch_california_housing()
Xtr,Xte,ytr,yte=train_test_split(housing.data,housing.target,test_size=.2,random_state=42)
model=DecisionTreeRegressor(max_depth=8,random_state=42).fit(Xtr,ytr)
print('MSE:',mean_squared_error(yte,model.predict(Xte)))
print(dict(zip(housing.feature_names,model.feature_importances_)))
# Requires access to download the California Housing dataset.
```

## Question 9. Write a Python program to: ●​ Load the Iris Dataset ●​ Tune the Decision Tree’s max_depth and min_samples_split using GridSearchCV ●​ Print the best parameters and the resulting model accuracy

Answer: Grid search tunes tree complexity using cross-validation. Report the best settings and the held-out accuracy.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
X,y=load_iris(return_X_y=True)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,stratify=y,random_state=42)
grid=GridSearchCV(DecisionTreeClassifier(random_state=42),
    {'max_depth':[2,3,4,5,None],'min_samples_split':[2,5,10]},cv=5)
grid.fit(Xtr,ytr)
print('Best parameters:',grid.best_params_)
print('Test accuracy:',grid.score(Xte,yte))
```
