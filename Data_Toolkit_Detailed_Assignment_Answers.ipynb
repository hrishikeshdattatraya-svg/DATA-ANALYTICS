## Theory Questions

### 1. What is NumPy, and why is it widely used in Python?

NumPy (Numerical Python) is a library for efficient numerical computing. Its ndarray stores values in compact, usually homogeneous arrays and supports fast mathematical operations. It is widely used in data science, machine learning, statistics, and scientific computing.

### 2. How does broadcasting work in NumPy?

Broadcasting lets NumPy perform operations on arrays of different but compatible shapes without manually copying data. It compares dimensions from the right; dimensions must match or one must equal 1. For example, adding 10 to an array adds 10 to every element.

### 3. What is a Pandas DataFrame?

A Pandas DataFrame is a two-dimensional, labeled table with rows and columns. Different columns can store different data types, such as names, dates, and numbers. It makes importing, filtering, cleaning, and analyzing datasets convenient.

### 4. Explain the use of the groupby() method in Pandas.

The groupby() method splits a DataFrame into groups based on one or more columns, applies an operation, and combines the results. For example, df.groupby("Department")["Salary"].mean() calculates the average salary in each department.

### 5. Why is Seaborn preferred for statistical visualizations?

Seaborn is a statistical visualization library built on Matplotlib. It provides attractive default styles and simple functions for distributions, correlations, and categorical comparisons. It also works directly with Pandas DataFrames, reducing plotting code.

### 6. What are the differences between NumPy arrays and Python lists?

Python lists can hold mixed types and are flexible general-purpose containers. NumPy arrays usually have one data type and use compact memory, making large numerical calculations faster. Arrays also support element-wise arithmetic directly, while lists do not.

### 7. What is a heatmap, and when should it be used?

A heatmap represents numerical values using colors in a grid. It is useful for visualizing correlation matrices, missing-value patterns, or measurements across two categories. Color intensity helps identify high, low, and unusual values quickly.

### 8. What does the term “vectorized operation” mean in NumPy?

A vectorized operation applies a calculation to an entire array without writing a Python loop for each element. For example, arr * 2 doubles every value. NumPy performs much of this work in optimized compiled code, improving speed and readability.

### 9. How does Matplotlib differ from Plotly?

Matplotlib is a flexible plotting library commonly used for static figures and detailed customization. Plotly focuses on interactive charts with hovering, zooming, and browser-based exploration. Both can create common plots, but Plotly is especially useful for interactive dashboards.

### 10. What is the significance of hierarchical indexing in Pandas?

Hierarchical indexing, or MultiIndex, allows multiple levels of row or column labels in a DataFrame. It helps organize higher-dimensional information in a two-dimensional table, such as sales indexed by region and year. Users can select, group, and reshape data by index level.

### 11. What is the role of Seaborn’s pairplot() function?

Seaborn’s pairplot() displays pairwise relationships between numeric variables in a dataset. It typically uses scatter plots off the diagonal and distributions on the diagonal. It is helpful for exploring correlations, clusters, and differences between categories.

### 12. What is the purpose of the describe() function in Pandas?

Pandas describe() produces summary statistics for columns. For numeric data it commonly shows count, mean, standard deviation, minimum, quartiles, and maximum. It is a quick first step for understanding data spread and detecting possible outliers.

### 13. Why is handling missing data important in Pandas?

Missing values can distort calculations, reduce usable observations, and cause errors in analysis or models. Pandas provides isna(), dropna(), and fillna() to identify or handle missing entries. The correct treatment depends on why values are missing and the analysis goal.

### 14. What are the benefits of using Plotly for data visualization?

Plotly provides interactive charts with tooltips, zooming, filtering through applications, and easy sharing as HTML. It supports 2D, 3D, and geographic plots. These features make it useful for exploratory analysis and interactive dashboards.

### 15. How does NumPy handle multidimensional arrays?

NumPy represents multidimensional data using the ndarray object, with attributes such as shape, ndim, and dtype. Arrays can represent matrices, image pixels, or higher-dimensional tensors. Indexing, slicing, reshaping, and axis-based calculations make them easy to manipulate.

### 16. What is the role of Bokeh in data visualization?

Bokeh is a Python library for creating interactive visualizations for web browsers. It supports tools such as hover, pan, and zoom and can connect plots to interactive applications. It is useful when users need to explore data visually rather than view only static images.

### 17. Explain the difference between apply() and map() in Pandas.

In Pandas, Series.map() transforms individual values using a function, dictionary, or Series mapping. Series.apply() also applies a function to values and can support more general operations; DataFrame.apply() operates along rows or columns. Choose map() for straightforward element-wise Series mappings.

### 18. What are some advanced features of NumPy?

Advanced NumPy features include broadcasting, fancy and Boolean indexing, linear algebra, random number generation, and fast Fourier transforms. It also supports reshaping, aggregation along axes, and memory-efficient views. These tools are valuable for large scientific datasets.

### 19. How does Pandas simplify time series analysis?

Pandas provides datetime indexes, date parsing, time-based filtering, resampling, and rolling-window calculations. For example, resample("ME").sum() can summarize daily values into monthly totals. These features simplify trend analysis and time-based reporting.

### 20. What is the role of a pivot table in Pandas?

A pivot table summarizes data by grouping rows and columns and applying an aggregation such as sum, count, or mean. In Pandas, pivot_table() can compare categories in a compact report. For example, it can show average sales by region and product.

### 21. Why is NumPy’s array slicing faster than Python’s list slicing?

NumPy slices often return views of existing array memory instead of copying all elements. This makes many slicing operations fast and memory-efficient. Python list slicing generally creates a new list, although actual performance depends on the operation and data size.

### 22. What are some common use cases for Seaborn?

Seaborn is commonly used for histograms, box plots, violin plots, scatter plots, correlation heatmaps, and pair plots. These charts help examine distributions, compare categories, identify outliers, and investigate relationships between variables.

## Practical Questions

### 1. How do you create a 2D NumPy array and calculate the sum of each row?

Use np.array() to create a matrix and sum(axis=1) to add values across each row.

```python
import numpy as np
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.sum(axis=1))  # [6 15]
```

### 2. Write a Pandas script to find the mean of a specific column in a DataFrame.

Select the required numeric column and call mean(). Missing values are ignored by default.

```python
import pandas as pd
df = pd.DataFrame({"Marks": [70, 80, 90]})
print(df["Marks"].mean())  # 80.0
```

### 3. Create a scatter plot using Matplotlib.

A scatter plot displays the relationship between two numerical variables using individual points.

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
y = [2, 4, 3, 7]
plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Scatter Plot")
plt.show()
```

### 4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?

Use DataFrame.corr() to compute correlations, then sns.heatmap() to display them with colors.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({"A": [1, 2, 3], "B": [2, 4, 5], "C": [6, 4, 2]})
corr = df.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```

### 5. Generate a bar plot using Plotly.

Plotly Express can create an interactive bar chart with category labels and numerical values.

```python
import plotly.express as px
fig = px.bar(x=["A", "B", "C"], y=[10, 15, 7],
             labels={"x": "Category", "y": "Value"})
fig.show()
```

### 6. Create a DataFrame and add a new column based on an existing column.

Assign a calculated Series to a new column name. The calculation is applied to every row.

```python
import pandas as pd
df = pd.DataFrame({"Price": [100, 200, 300]})
df["Tax"] = df["Price"] * 0.18
print(df)
```

### 7. Write a program to perform element-wise multiplication of two NumPy arrays.

The * operator multiplies corresponding elements of compatible NumPy arrays.

```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a * b)  # [4 10 18]
```

### 8. Create a line plot with multiple lines using Matplotlib.

Call plot() once for each series and use a legend to distinguish the lines.

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
plt.plot(x, [2, 4, 6, 8], label="Series A")
plt.plot(x, [1, 3, 5, 7], label="Series B")
plt.legend()
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

### 9. Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold.

Boolean indexing keeps only rows that satisfy the condition.

```python
import pandas as pd
df = pd.DataFrame({"Name": ["A", "B", "C"], "Score": [45, 75, 90]})
print(df[df["Score"] > 60])
```

### 10. Create a histogram using Seaborn to visualize a distribution.

A histogram groups numeric values into bins and shows their frequency.

```python
import seaborn as sns
import matplotlib.pyplot as plt
values = [10, 12, 12, 14, 15, 15, 15, 18, 20]
sns.histplot(values, bins=5, kde=True)
plt.xlabel("Value")
plt.show()
```

### 11. Perform matrix multiplication using NumPy.

Use the @ operator or np.matmul() for matrix multiplication; this differs from element-wise multiplication.

```python
import numpy as np
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(a @ b)  # [[19 22], [43 50]]
```

### 12. Use Pandas to load a CSV file and display its first 5 rows.

read_csv() loads a CSV file into a DataFrame, and head() displays the first five rows by default.

```python
import pandas as pd
df = pd.read_csv("data.csv")  # File must exist
print(df.head())
```

### 13. Create a 3D scatter plot using Plotly.

Plotly Express scatter_3d() displays points using three numerical axes and interactive rotation.

```python
import plotly.express as px
fig = px.scatter_3d(x=[1, 2, 3], y=[3, 2, 5], z=[2, 4, 6],
                    labels={"x": "X", "y": "Y", "z": "Z"})
fig.show()
```
