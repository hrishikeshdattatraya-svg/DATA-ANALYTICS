## Question 1. What is hypothesis testing in statistics?

Hypothesis testing is a statistical method used to decide whether sample evidence supports a claim about a population. We begin with a null hypothesis (H₀), choose an alternative hypothesis (H₁), and calculate a test statistic and p-value from sample data. The p-value indicates how unusual the observed result would be if H₀ were true. If the p-value is at or below a chosen significance level (often 0.05), we reject H₀; otherwise, we fail to reject it. For example, a manufacturer may test whether the average weight of its packets differs from 500 g.

## Question 2. What is the null hypothesis, and how does it differ from the alternative hypothesis?

The null hypothesis (H₀) is the baseline claim, usually stating that there is no effect, no difference, or that a population parameter equals a specified value. The alternative hypothesis (H₁ or Hₐ) states that an effect or difference exists. For example, to test a claimed average battery life of 10 hours, H₀: μ = 10 and H₁: μ ≠ 10. We assess the evidence against H₀; failing to reject H₀ does not prove it is true. An alternative may be two-sided (≠) or one-sided (> or <).

## Question 3. Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.

The significance level, denoted α, is the maximum probability of a Type I error that the researcher is willing to accept under the testing procedure. A common choice is α = 0.05. The decision rule is: reject H₀ when p ≤ α, and otherwise fail to reject H₀. For example, p = 0.03 is statistically significant at the 5% level, while p = 0.08 is not. A smaller α makes rejection harder and reduces false positives, but may also reduce statistical power. Statistical significance does not necessarily mean practical importance.

## Question 4. What are Type I and Type II errors? Give examples of each.

A Type I error occurs when we reject a true null hypothesis (a false positive). Its probability is controlled by α. For example, a medical screening test may wrongly indicate a disease in a healthy person. A Type II error occurs when we fail to reject a false null hypothesis (a false negative); its probability is β. For example, a test may fail to detect a real improvement caused by a new medicine. Statistical power is 1 − β, the probability of detecting an effect when one truly exists. Increasing sample size often improves power.

## Question 5. What is the difference between a Z-test and a T-test? Explain when to use each.

Both tests compare a sample statistic with a hypothesized population value. A Z-test uses the standard normal distribution, typically when the population standard deviation σ is known and the sampling assumptions are satisfied. A one-sample Z-statistic is z = (x̄ − μ₀)/(σ/√n). A T-test uses the Student t-distribution and the sample standard deviation s when σ is unknown: t = (x̄ − μ₀)/(s/√n), with n − 1 degrees of freedom. For small samples, the t-distribution accounts for extra uncertainty; for large samples it approaches the normal distribution. Both require appropriate independence and distribution assumptions.

## Question 6. Write a Python program to generate a binomial distribution with n=10 and p=0.5, then plot its histogram.

A binomial random variable counts successes in 10 independent trials, each having success probability 0.5. The following reproducible simulation generates 1,000 observations and plots the frequency of each possible number of successes.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
samples = rng.binomial(n=10, p=0.5, size=1000)
plt.hist(samples, bins=np.arange(-0.5, 11.5, 1), edgecolor="black")
plt.xticks(range(11))
plt.xlabel("Number of successes")
plt.ylabel("Frequency")
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.show()
print("Sample mean:", round(samples.mean(), 2))
```

Expected interpretation: The histogram is roughly symmetric and concentrated around 5 successes. The theoretical mean is np = 5. The exact sample mean varies with the generated observations.

## Question 7. Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results.

The question provides 36 measurements but does not specify a hypothesized mean or known population standard deviation. For illustration, we test H₀: μ = 50 versus H₁: μ ≠ 50, assuming the population standard deviation is known to be σ = 1.0. Without this assumption, a one-sample t-test is more appropriate.

```python
import numpy as np
from scipy.stats import norm

sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
               50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
               50.3, 50.4, 50.0, 49.7, 50.5, 49.9]
mu0, sigma, alpha = 50, 1.0, 0.05
mean = np.mean(sample_data)
z = (mean - mu0) / (sigma / np.sqrt(len(sample_data)))
p_value = 2 * norm.sf(abs(z))
print(f"Mean: {mean:.4f}, Z: {z:.4f}, p-value: {p_value:.4f}")
print("Reject H0" if p_value <= alpha else "Fail to reject H0")
```

Output: Mean = 50.0889; Z = 0.5333; p-value = 0.5938. At α = 0.05, fail to reject H₀.

## Question 8. Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.

We generate 100 observations from a normal distribution and estimate the population mean. Because the population standard deviation is treated as unknown, we use a t-based 95% confidence interval.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t

rng = np.random.default_rng(42)
data = rng.normal(loc=50, scale=10, size=100)
n = len(data)
mean = np.mean(data)
se = np.std(data, ddof=1) / np.sqrt(n)
margin = t.ppf(0.975, df=n - 1) * se
print(f"Mean: {mean:.2f}")
print(f"95% CI: ({mean - margin:.2f}, {mean + margin:.2f})")
plt.hist(data, bins=12, edgecolor="black")
plt.axvline(mean, color="red", label="Sample mean")
plt.title("Simulated Normal Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

Interpretation: The interval gives a range of plausible population means under the model. Across repeated samples, approximately 95% of intervals built using this method would contain the true population mean.

## Question 9. Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.

A Z-score measures how far a value is from the dataset mean in standard deviation units: z = (x − x̄)/s. Positive Z-scores are above the mean, negative scores are below it, and zero represents the mean. A score of +2 is two standard deviations above the mean. The function below handles constant-valued datasets by raising an error.

```python
import numpy as np
import matplotlib.pyplot as plt

def calculate_z_scores(values):
    values = np.asarray(values, dtype=float)
    if values.size < 2:
        raise ValueError("At least two values are required")
    sd = np.std(values, ddof=1)
    if sd == 0:
        raise ValueError("Standard deviation must not be zero")
    return (values - np.mean(values)) / sd

data = [10, 12, 13, 15, 18, 20, 22, 25, 27, 30]
z_scores = calculate_z_scores(data)
print("Z-scores:", np.round(z_scores, 2))
plt.hist(z_scores, bins=6, edgecolor="black")
plt.axvline(0, color="red", linestyle="--")
plt.xlabel("Z-score")
plt.ylabel("Frequency")
plt.title("Histogram of Standardized Data")
plt.show()
```

Interpretation: Standardization changes the scale but preserves the relative ordering and overall shape of the distribution. The standardized sample has mean approximately 0 and sample standard deviation 1.
