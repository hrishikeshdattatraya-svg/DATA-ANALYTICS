## Question 1. Difference between descriptive and inferential statistics

Descriptive statistics summarize and present the data that have already been collected. Common methods include the mean, median, percentages, tables and charts. For example, calculating the average marks of all 40 students in one class describes that class.

Inferential statistics use a sample to draw conclusions about a larger population. They involve uncertainty and methods such as confidence intervals and hypothesis tests. For example, surveying 200 voters to estimate the preferences of all voters in a city is an inference. The key difference is that descriptive statistics describe observed data, while inferential statistics generalize beyond it.

## Question 2. Sampling; random and stratified sampling

Sampling is the process of selecting a smaller group (sample) from a larger population to study its characteristics. It saves time and cost when collecting information from everyone is impractical.

In simple random sampling, every member has an equal chance of selection; for example, randomly selecting 50 roll numbers from a college list. In stratified sampling, the population is first divided into meaningful groups (strata), such as academic years, and random samples are drawn from each group. Stratified sampling helps ensure important groups are represented and can improve precision when groups differ.

## Question 3. Mean, median and mode; their importance

Mean is the arithmetic average: sum of all observations divided by their count. Median is the middle value after sorting the data (or the average of the two middle values). Mode is the most frequently occurring value; a dataset may have more than one mode.

These measures describe a typical value and make datasets easier to compare. The mean uses every observation but is affected by extreme values; the median is more robust to outliers; and the mode is useful for identifying the most common category or value. For example, for 2, 3, 3, 4, 18, the mean is 6, the median is 3, and the mode is 3.

## Question 4. Skewness and kurtosis; meaning of positive skew

Skewness measures the asymmetry of a distribution. A symmetric distribution has skewness near zero; a positively skewed (right-skewed) distribution has a longer tail toward larger values, while a negatively skewed distribution has a longer left tail.

Positive skew often occurs when most observations are relatively small but a few are very large, as with household incomes. In such cases the mean is often greater than the median. Kurtosis describes the heaviness of distribution tails and the tendency for extreme observations relative to a normal distribution. High excess kurtosis indicates heavier tails, not simply a taller peak.

## Question 5. Python program for mean, median and mode

The statistics module computes the mean and median directly. Since three numbers occur equally often in this dataset, multimode() correctly returns all modes rather than selecting just one.

````python
import statistics as stats
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
print("Mean:", round(stats.mean(numbers), 2))
print("Median:", stats.median(numbers))
print("Modes:", stats.multimode(numbers))
````

**Expected output:**

```text
Mean: 19.60
Median: 19
Modes: [12, 19, 24]
```

## Question 6. Covariance and correlation coefficient

Covariance shows whether two variables tend to increase or decrease together. Pearson's correlation standardizes the relationship to a value between −1 and +1. The code below uses sample covariance (ddof=1) and Pearson correlation.

The positive covariance and correlation close to +1 indicate a very strong positive linear relationship between these two datasets.

````python
import numpy as np
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]
covariance = np.cov(list_x, list_y)[0, 1]
correlation = np.corrcoef(list_x, list_y)[0, 1]
print("Sample covariance:", round(covariance, 2))
print("Pearson correlation:", round(correlation, 4))
````

**Expected output:**

```text
Sample covariance: 275.0
Pearson correlation: 0.9959
```

## Question 7. Boxplot and identification of outliers

A boxplot displays the median, quartiles, whiskers and possible outliers. Under the standard 1.5 × IQR rule, observations below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are marked as outliers. The program uses the same quartile method as Matplotlib's default boxplot.

For this list, Q1 = 17.25, Q3 = 23.25 and IQR = 6. The upper fence is 32.25, so 35 is an outlier; the lower fence is 8.25 and there are no lower outliers.

````python
import numpy as np
import matplotlib.pyplot as plt
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
q1, q3 = np.percentile(data, [25, 75])
iqr = q3 - q1
outliers = [v for v in data if v < q1 - 1.5*iqr or v > q3 + 1.5*iqr]
print("Q1:", q1, "Q3:", q3, "IQR:", iqr)
print("Outliers:", outliers)
plt.boxplot(data)
plt.ylabel("Values")
plt.title("Boxplot of numeric data")
plt.show()
````

**Expected output:**

```text
Q1: 17.25 Q3: 23.25 IQR: 6.0
Outliers: [35]
```

## Question 8. Advertising spend and daily sales: covariance and correlation

Covariance can show whether advertising spend and daily sales move in the same direction, but its size depends on the measurement units. Pearson correlation gives a unit-free measure of the strength of their linear association.

The calculated correlation is about 0.9936, indicating a very strong positive relationship in these five observations. This association alone does not prove that advertising causes the increase in sales; other factors may influence sales.

````python
import numpy as np
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
correlation = np.corrcoef(advertising_spend, daily_sales)[0, 1]
print("Correlation:", round(correlation, 4))
````

**Expected output:**

```text
Correlation: 0.9936
```

## Question 9. Customer satisfaction: summary statistics and histogram

For satisfaction scores from 1 to 10, I would calculate the mean and median to summarize typical satisfaction, the mode for the most frequent score, and standard deviation to measure variability. Minimum, maximum and score frequencies also help identify low ratings or concentration around high ratings.

A histogram shows how many customers selected scores in each interval, making the distribution, spread and possible skew easy to see. A boxplot could additionally highlight the median and possible outliers. The following code creates a histogram with one bin per integer score.

````python
import numpy as np
import matplotlib.pyplot as plt
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
print("Mean:", round(np.mean(survey_scores), 2))
print("Median:", np.median(survey_scores))
print("Sample standard deviation:", round(np.std(survey_scores, ddof=1), 2))
plt.hist(survey_scores, bins=np.arange(0.5, 11.5, 1), edgecolor="black")
plt.xticks(range(1, 11))
plt.xlabel("Satisfaction score")
plt.ylabel("Number of customers")
plt.title("Customer satisfaction distribution")
plt.show()
````

**Expected output:**

```text
Mean: 7.33
Median: 7.0
Sample standard deviation: 1.63
```
