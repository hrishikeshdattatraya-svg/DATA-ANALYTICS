## Part A — Theory Questions

### 1. What is Object-Oriented Programming (OOP)?

Object-Oriented Programming is a programming approach that organizes code around objects containing data (attributes) and behavior (methods). Its main principles are encapsulation, abstraction, inheritance, and polymorphism. For example, a Student object can store a name and marks and provide a calculate_grade() method.

### 2. What is a class in OOP?

A class is a blueprint used to create objects. It defines the attributes and methods that its objects can have. For example, a Car class may define color and speed attributes along with start() and stop() methods. Each object created from it can hold different values.

### 3. What is an object in OOP?

An object is an instance of a class created during program execution. It has its own state and can use methods defined by the class. For example, car1 = Car() creates an object of the Car class; its attributes can describe a specific vehicle.

### 4. What is the difference between abstraction and encapsulation?

Abstraction hides implementation complexity and exposes only essential operations, such as a payment interface that offers pay(). Encapsulation groups data with methods and controls how internal state is accessed, such as a bank account allowing balance changes only through deposit() and withdraw(). Abstraction focuses on what is available; encapsulation focuses on protecting and organizing state.

### 5. What are dunder methods in Python?

Dunder (double-underscore) methods are special methods such as __init__, __str__, __len__, and __add__. Python calls them for built-in operations including object creation, printing, length checks, and addition. They let user-defined classes behave naturally with Python syntax.

### 6. Explain the concept of inheritance in OOP.

Inheritance allows a child class to reuse and extend behavior from a parent class. For example, Dog can inherit from Animal and override speak(). It reduces repeated code and represents “is-a” relationships. Python supports single, multiple, and multilevel inheritance.

### 7. What is polymorphism in OOP?

Polymorphism means different objects can respond to the same operation in different ways. For instance, Dog.speak() can return “Bark” while Cat.speak() returns “Meow”. Code can call speak() on either object without needing separate logic for every animal type.

### 8. How is encapsulation achieved in Python?

Python uses class methods, properties, and naming conventions to manage access to attributes. A leading underscore marks an attribute as non-public by convention; double underscores trigger name mangling, which discourages accidental access but is not strict security. Properties can validate changes while preserving a simple attribute interface.

### 9. What is a constructor in Python?

A constructor initializes an object when it is created. In everyday Python code, __init__ is the initializer used to set instance attributes, while __new__ creates the instance. For example, __init__(self, name) can store self.name = name.

### 10. What are class and static methods in Python?

A class method uses @classmethod and receives the class as cls; it is useful for alternative constructors or class-wide operations. A static method uses @staticmethod and receives neither self nor cls automatically; it is useful for helper functions logically related to the class.

### 11. What is method overloading in Python?

Method overloading usually means defining multiple methods with the same name but different parameters. Python does not support traditional signature-based overloading within one class: later definitions replace earlier ones. Similar behavior can be implemented with default parameters, *args, or type-based dispatch.

### 12. What is method overriding in OOP?

Overriding occurs when a child class provides its own implementation of an inherited method. For example, Dog.speak() can replace Animal.speak() for Dog objects. It supports runtime polymorphism and lets subclasses customize behavior.

### 13. What is a property decorator in Python?

The @property decorator lets a method be accessed like an attribute. A matching @name.setter can validate assigned values. For example, a temperature property can reject values below an allowed minimum while allowing object.temperature syntax.

### 14. Why is polymorphism important in OOP?

Polymorphism makes code flexible and reusable. One function can work with different objects that provide the same method, reducing conditional statements and coupling. It also makes systems easier to extend because new subclasses can be introduced without rewriting calling code.

### 15. What is an abstract class in Python?

An abstract class defines a common interface and can require subclasses to implement particular methods. Python provides ABC and @abstractmethod in the abc module. A class with unimplemented abstract methods cannot normally be instantiated, helping enforce consistent behavior.

### 16. What are the advantages of OOP?

OOP improves organization by grouping related data and behavior. Inheritance and composition promote reuse; encapsulation helps protect object state; and polymorphism supports flexible interfaces. These features can make large applications easier to maintain, test, and extend when classes are designed carefully.

### 17. What is the difference between a class variable and an instance variable?

A class variable is defined in the class body and is shared through the class unless shadowed. An instance variable is typically assigned using self and belongs to a particular object. For example, Student.school can be shared, while self.name differs for each student.

### 18. What is multiple inheritance in Python?

Multiple inheritance allows one class to inherit from more than one parent, such as class SmartPhone(Camera, Phone). Python resolves attribute and method lookup using its method resolution order (MRO). It can combine behaviors but should be designed carefully to avoid ambiguity.

### 19. Explain the purpose of __str__ and __repr__ methods in Python.

__str__ returns a readable description intended for users and is used by str() and print(). __repr__ returns a representation mainly intended for debugging, ideally unambiguous and sometimes valid Python code. If __str__ is missing, Python can fall back to __repr__.

### 20. What is the significance of the super() function in Python?

super() gives access to methods in the next class in the method resolution order. It is commonly used to call a parent initializer or extend inherited behavior without directly naming the parent. It is particularly helpful in cooperative multiple inheritance.

### 21. What is the significance of the __del__ method in Python?

__del__ is a finalizer that Python may call when an object is being destroyed. Its execution timing is not guaranteed, so it should not be relied on for essential cleanup. Context managers (with statements) or explicit close() methods are better for files, connections, and other resources.

### 22. What is the difference between @staticmethod and @classmethod in Python?

@staticmethod creates a function in a class namespace without automatically receiving an object or class argument. @classmethod automatically receives cls and can inspect or create instances of the actual subclass. Use static methods for independent utilities and class methods for class-aware operations.

### 23. How does polymorphism work in Python with inheritance?

A parent reference can refer to an instance of a child class. When an overridden method is called, Python selects the implementation on the actual object at runtime. For example, iterating over Dog and Cat objects and calling speak() produces each animal’s own sound.

### 24. What is method chaining in Python OOP?

Method chaining means calling several methods consecutively on an object, such as builder.set_name("A").set_age(20).build(). Methods commonly return self to allow the next call. It can make configuration concise, although overly long chains may reduce readability.

### 25. What is the purpose of the __call__ method in Python?

The __call__ method makes an instance callable like a function. If an object defines __call__(self, x), then obj(x) executes that method. This is useful for callable configuration objects, counters, and reusable processing components.

### 2. Abstract Shape with Circle and Rectangle

```python
from abc import ABC, abstractmethod
from math import pi

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width, self.height = width, height
    def area(self):
        return self.width * self.height

print(round(Circle(3).area(), 2))
print(Rectangle(4, 5).area())
```

Sample output: 28.27 | 20

Explanation: This program demonstrates the requested OOP concept using classes, methods, and an example object or method call.

### 3. Multilevel inheritance: Vehicle, Car, ElectricCar

```python
class Vehicle:
    def __init__(self, type):
        self.type = type

class Car(Vehicle):
    def __init__(self, brand):
        super().__init__("Car")
        self.brand = brand

class ElectricCar(Car):
    def __init__(self, brand, battery):
        super().__init__(brand)
        self.battery = battery

car = ElectricCar("Tesla", "75 kWh")
print(car.type, car.brand, car.battery)
```

### 4. Bird polymorphism: Sparrow and Penguin

```python
class Bird:
    def fly(self):
        print("Bird flight behavior")

class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies.")

class Penguin(Bird):
    def fly(self):
        print("Penguin cannot fly.")

for bird in (Sparrow(), Penguin()):
    bird.fly()
```

Sample output: Sparrow flies. | Penguin cannot fly.

Explanation: This program demonstrates the requested OOP concept using classes, methods, and an example object or method call.

### 5. BankAccount encapsulation

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        self.__balance += amount
    def withdraw(self, amount):
        if amount <= 0 or amount > self.__balance:
            raise ValueError("Invalid withdrawal")
        self.__balance -= amount
    def check_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
account.withdraw(200)
print(account.check_balance())
```

Sample output: 1300

Explanation: This program demonstrates the requested OOP concept using classes, methods, and an example object or method call.

### 9. Fraction with __str__

### 10. Vector operator overloading using __add__

```python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    def __str__(self):
        return f"({self.x}, {self.y})"

print(Vector(2, 3) + Vector(4, 5))
```

### 18. House and Mansion inheritance
