## Question 1. What is Logistic Regression, and how does it differ from Linear Regression?

Answer: Logistic Regression is a supervised classification algorithm that models the probability of a class using a linear score passed through a sigmoid function. Linear Regression predicts unrestricted continuous values, whereas binary Logistic Regression predicts probabilities between 0 and 1. A decision threshold converts the predicted probability into a class label.

## Question 2. Explain the role of the Sigmoid function in Logistic Regression.

Answer: The sigmoid function is σ(z) = 1/(1 + e^(−z)). It maps any real-valued linear score to a probability between 0 and 1. When z = 0 the output is 0.5; large positive scores approach 1 and large negative scores approach 0. Logistic Regression learns coefficients by optimizing a classification loss, typically log loss.

## Question 3. What is Regularization in Logistic Regression and why is it needed?

Answer: Regularization penalizes large model coefficients to reduce overfitting and improve generalization. L2 regularization shrinks weights smoothly, while L1 can drive some weights to zero and perform feature selection. The inverse strength parameter C in scikit-learn controls the penalty: smaller C means stronger regularization.

## Question 4. What are some common evaluation metrics for classification models, and why are they important?

Answer: Accuracy is the fraction of correct predictions, but it can be misleading for imbalanced classes. Precision measures the proportion of predicted positives that are correct, recall measures the proportion of actual positives detected, and F1 balances precision and recall. ROC-AUC assesses ranking across thresholds; PR-AUC is especially informative for rare positive classes.

## Question 5. Write a Python program that loads a CSV file into a Pandas DataFrame, splits into train/test sets, trains a Logistic Regression model, and prints its accuracy.​ (Use Dataset from sklearn package)

Answer: The Breast Cancer dataset is bundled with scikit-learn. Save it as CSV, read it using pandas, split the observations and fit a classifier.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
b = load_breast_cancer(as_frame=True)
df = b.frame
df.to_csv('breast_cancer.csv', index=False)
df = pd.read_csv('breast_cancer.csv')
X, y = df.drop(columns='target'), df['target']
Xtr,Xte,ytr,yte = train_test_split(X,y,test_size=.2,stratify=y,random_state=42)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000))
model.fit(Xtr,ytr)
print('Accuracy:', model.score(Xte,yte))
```

## Question 6. Write a Python program to train a Logistic Regression model using L2 regularization (Ridge) and print the model coefficients and accuracy. (Use Dataset from sklearn package)

Answer: L2-regularized logistic regression penalizes squared coefficient magnitude. Print learned coefficients and held-out accuracy after standardizing the input features.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
X,y = load_breast_cancer(return_X_y=True)
Xtr,Xte,ytr,yte = train_test_split(X,y,test_size=.2,stratify=y,random_state=42)
model = make_pipeline(StandardScaler(), LogisticRegression(penalty='l2',C=1,max_iter=2000))
model.fit(Xtr,ytr)
print('Coefficients:', model[-1].coef_)
print('Accuracy:', model.score(Xte,yte))
```

## Question 7. Write a Python program to train a Logistic Regression model for multiclass classification using multi_class='ovr' and print the classification report.​ (Use Dataset from sklearn package)

Answer: One-vs-rest trains one binary classifier for each class. The following implementation uses OneVsRestClassifier, which is a current and explicit alternative to the deprecated multi_class="ovr" option.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report
X,y = load_iris(return_X_y=True)
Xtr,Xte,ytr,yte = train_test_split(X,y,test_size=.3,stratify=y,random_state=42)
model = make_pipeline(StandardScaler(), OneVsRestClassifier(LogisticRegression(max_iter=1000)))
model.fit(Xtr,ytr)
print(classification_report(yte,model.predict(Xte)))
```

## Question 8. Write a Python program to apply GridSearchCV to tune C and penalty hyperparameters for Logistic Regression and print the best parameters and validation accuracy. (Use Dataset from sklearn package)

Answer: GridSearchCV evaluates combinations of C and penalty using cross-validation on training data. The test set remains untouched until the best model is selected.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
X,y = load_breast_cancer(return_X_y=True)
Xtr,Xte,ytr,yte = train_test_split(X,y,test_size=.2,stratify=y,random_state=42)
pipe = make_pipeline(StandardScaler(),LogisticRegression(solver='liblinear',max_iter=2000))
search = GridSearchCV(pipe,{'logisticregression__C':[.01,.1,1,10],
    'logisticregression__penalty':['l1','l2']},cv=5,scoring='accuracy')
search.fit(Xtr,ytr)
print('Best:',search.best_params_,'CV accuracy:',search.best_score_)
print('Test accuracy:',search.score(Xte,yte))
```

## Question 9. Write a Python program to standardize the features before training Logistic Regression and compare the model's accuracy with and without scaling. (Use Dataset from sklearn package)

Answer: Compare the same logistic model with and without feature scaling on a fixed train-test split. Scaling often improves convergence and the fairness of coefficient regularization.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
X,y = load_breast_cancer(return_X_y=True)
Xtr,Xte,ytr,yte = train_test_split(X,y,test_size=.2,stratify=y,random_state=42)
plain = LogisticRegression(max_iter=5000)
scaled = make_pipeline(StandardScaler(),LogisticRegression(max_iter=5000))
for name, model in [('Unscaled',plain),('Scaled',scaled)]:
    model.fit(Xtr,ytr)
    print(name, 'accuracy:',model.score(Xte,yte))
```
