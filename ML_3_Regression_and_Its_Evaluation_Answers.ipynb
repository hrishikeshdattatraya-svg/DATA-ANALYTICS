## Question 1. What is Simple Linear Regression?

Answer: Simple Linear Regression models a continuous response Y using one predictor X with the equation Y = β₀ + β₁X + ε. The intercept β₀ is the predicted outcome at X = 0, and β₁ is the expected change in Y for one unit of X. Ordinary least squares estimates coefficients by minimizing squared residuals. Example: predicting sales from advertising spending.

## Question 2. What are the key assumptions of Simple Linear Regression?

Answer: Key assumptions for standard linear-regression inference include a linear mean relationship, independent errors, constant error variance (homoscedasticity), and no perfect collinearity. Residual normality is useful for small-sample confidence intervals and tests, but is not required merely to fit predictions. Check scatterplots, residual plots and domain knowledge.

## Question 3. What is heteroscedasticity, and why is it important to address in regression models?

Answer: Heteroscedasticity means the error variance changes across predictor values or fitted outcomes. It can make ordinary least-squares standard errors and hypothesis tests unreliable, although coefficient estimates can remain unbiased under other assumptions. Investigate residual-versus-fitted plots, consider transforming the response, weighted least squares or heteroscedasticity-robust standard errors.

## Question 4. What is Multiple Linear Regression?

Answer: Multiple Linear Regression predicts a continuous response from two or more explanatory variables: Y = β₀ + β₁X₁ + β₂X₂ + ... + ε. Each coefficient describes a predictor’s association with the outcome while holding other included predictors fixed. For example, house price may depend on floor area, number of rooms and neighborhood.

## Question 5. What is polynomial regression, and how does it differ from linear regression?

Answer: Polynomial Regression includes powers of predictors, such as Y = β₀ + β₁X + β₂X² + ε, allowing curved relationships. It remains linear in the coefficients even though the fitted curve is nonlinear in X. Degree selection should use cross-validation because high-degree polynomials can overfit.

## Question 6. Implement a Python program to fit a Simple Linear Regression model to the following sample data: ●​ X = [1, 2, 3, 4, 5] ●​ Y = [2.1, 4.3, 6.1, 7.9, 10.2] Plot the regression line over the data points.

Answer: Fit a line to the given points, print its coefficients and plot predicted values against observed data.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
X = np.array([1,2,3,4,5]).reshape(-1,1)
y = np.array([2.1,4.3,6.1,7.9,10.2])
model = LinearRegression().fit(X,y)
print('Intercept:', model.intercept_)
print('Slope:', model.coef_[0])
plt.scatter(X.ravel(), y, label='Observed')
plt.plot(X.ravel(), model.predict(X), color='red', label='Fitted line')
plt.legend(); plt.xlabel('X'); plt.ylabel('Y'); plt.show()
```

## Question 7. Fit a Multiple Linear Regression model on this sample data: ●​ Area = [1200, 1500, 1800, 2000] ●​ Rooms = [2, 3, 3, 4] ●​ Price = [250000, 300000, 320000, 370000] Check for multicollinearity using VIF and report the results.​

Answer: Fit a multiple regression using area and rooms. VIF values quantify how strongly a predictor can be explained by the other predictors; elevated VIF indicates unstable individual coefficient estimates. This tiny dataset supports illustration, not robust inference.

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.linear_model import LinearRegression
X = pd.DataFrame({'Area':[1200,1500,1800,2000],
                  'Rooms':[2,3,3,4]})
y = [250000,300000,320000,370000]
model = LinearRegression().fit(X,y)
print('Coefficients:', model.coef_)
X_const = sm.add_constant(X)
for i, name in enumerate(X_const.columns):
    if name != 'const':
        print(name, 'VIF:', variance_inflation_factor(X_const.values, i))
```

## Question 8. Implement polynomial regression on the following data: ●​ X = [1, 2, 3, 4, 5] 3 ●​ Y = [2.2, 4.8, 7.5, 11.2, 14.7] Fit a 2nd-degree polynomial and plot the resulting curve.

Answer: Create degree-two features, fit a linear estimator and draw the smooth polynomial curve.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
X = np.array([1,2,3,4,5]).reshape(-1,1)
y = np.array([2.2,4.8,7.5,11.2,14.7])
poly = PolynomialFeatures(degree=2, include_bias=False)
model = LinearRegression().fit(poly.fit_transform(X), y)
xx = np.linspace(1,5,100).reshape(-1,1)
plt.scatter(X,y); plt.plot(xx, model.predict(poly.transform(xx)), color='red')
plt.title('Second-degree polynomial fit'); plt.show()
print('Coefficients:', model.coef_)
```

## Question 9. Create a residuals plot for a regression model trained on this data: ●​ X = [10, 20, 30, 40, 50] ●​ Y = [15, 35, 40, 50, 65] Assess heteroscedasticity by examining the spread of residuals.

Answer: Residuals are observed values minus predictions. Plot them against fitted values and look for widening or narrowing spread. With only five observations, heteroscedasticity cannot be assessed reliably.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
X = np.array([10,20,30,40,50]).reshape(-1,1)
y = np.array([15,35,40,50,65])
model = LinearRegression().fit(X,y)
fitted = model.predict(X)
residuals = y - fitted
print('Residuals:', residuals)
plt.scatter(fitted,residuals); plt.axhline(0,color='red',linestyle='--')
plt.xlabel('Fitted'); plt.ylabel('Residuals'); plt.show()
```
