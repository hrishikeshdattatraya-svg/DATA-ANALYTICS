## Part A — Theory

### 1. What is a RESTful API?

A RESTful API is a web interface that follows REST principles. It exposes resources through URLs, uses standard HTTP methods, and generally exchanges JSON. Each request contains the information needed to process it (stateless communication).

### 2. Explain the concept of API specification.

An API specification is a contract describing endpoints, request methods, parameters, authentication, response formats, status codes, and errors. OpenAPI is a common format for documenting and testing this contract.

### 3. What is Flask, and why is it popular for building APIs?

Flask is a lightweight Python web framework. It is popular because its routing is simple, it is flexible, and developers can add database, authentication, and validation libraries as needed.

### 4. What is routing in Flask?

Routing maps a URL path and HTTP method to a Python view function. For example, @app.get('/hello') makes the hello function handle GET /hello.

### 5. How do you create a simple Flask application?

Install Flask, create a Flask instance, define a route, and run the development server. See Practical 1 for a complete example.

### 6. What are HTTP methods used in RESTful APIs?

GET retrieves resources; POST creates or submits data; PUT replaces a resource; PATCH partially updates it; DELETE removes it. HEAD retrieves headers and OPTIONS describes supported communication options.

### 7. What is the purpose of the @app.route() decorator in Flask?

It registers a URL rule and associates it with a view function. The methods argument restricts the accepted HTTP methods.

### 8. What is the difference between GET and POST HTTP methods?

GET requests retrieve information and should not change server state; parameters commonly appear in the URL. POST submits data in the request body and may create resources or trigger processing.

### 9. How do you handle errors in Flask APIs?

Return appropriate HTTP status codes and consistent JSON error messages. Use abort() for HTTP errors, @app.errorhandler() for custom responses, and log unexpected exceptions without exposing sensitive details.

### 10. How do you connect Flask to a SQL database?

Configure a database connection URL, initialize an ORM such as Flask-SQLAlchemy, define models, and execute queries inside Flask's application context. Protect credentials through environment variables.

### 11. What is the role of Flask-SQLAlchemy?

Flask-SQLAlchemy integrates SQLAlchemy with Flask. It helps configure database connections, define models, and manage sessions and queries.

### 12. What are Flask blueprints, and how are they useful?

Blueprints group related routes and handlers into reusable modules. They improve maintainability in larger applications by separating features such as users and products.

### 13. What is the purpose of Flask's request object?

The request object provides incoming HTTP information such as method, headers, query parameters (request.args), form fields (request.form), and JSON bodies (request.get_json()).

### 14. How do you create a RESTful API endpoint using Flask?

Define a route for a resource, specify its HTTP method, process the request, and return JSON with a suitable status code. See Practical 14.

### 15. What is the purpose of Flask's jsonify() function?

jsonify() serializes supported Python values to JSON and creates a Flask Response with the application/json content type.

### 16. Explain Flask’s url_for() function.

url_for() builds URLs using endpoint function names instead of hardcoded paths. For example, url_for('home') resolves the route registered for home.

### 17. How does Flask handle static files (CSS, JavaScript, etc.)?

Flask serves files from its static/ directory during development. Reference them with url_for('static', filename='style.css'). A production web server or CDN may serve these assets.

### 18. What is an API specification, and how does it help in building a Flask API?

An API specification defines a shared, testable contract for Flask endpoints. It supports documentation, client generation, validation, and consistent collaboration. OpenAPI is widely used.

### 19. What are HTTP status codes, and why are they important in a Flask API?

HTTP status codes communicate the outcome of a request: 200 OK, 201 Created, 204 No Content, 400 Bad Request, 401 Unauthorized, 403 Forbidden, 404 Not Found, and 500 Internal Server Error. Clients use them to handle results reliably.

### 20. How do you handle POST requests in Flask?

Register a route with methods=['POST'] or @app.post(), read and validate request data, perform the action, and return a response. For JSON use request.get_json(silent=True).

### 21. How would you secure a Flask API?

Use HTTPS, strong authentication, server-side authorization, input validation, parameterized database queries, rate limiting, secure secrets, restricted CORS, and careful error handling. Keep dependencies updated.

### 22. What is the significance of the Flask-RESTful extension?

Flask-RESTful provides resource classes, method-based handlers, and helpers for REST APIs. It is optional; modern Flask can also build REST APIs directly.

### 23. What is the role of Flask’s session object?

Flask's session stores small user-specific values across requests using a cryptographically signed cookie by default. Its contents are not encrypted, so do not store secrets in it. Set a strong SECRET_KEY.

## Part B — Practical

Each example is a standalone snippet unless noted. Install Flask with: pip install flask

### 1. How do you create a basic Flask application?

```python
from flask import Flask
app = Flask(__name__)

@app.get('/')
def home():
    return 'Hello, Flask!'

if __name__ == '__main__':
    app.run(debug=True)
```

Install with pip install flask, then run python app.py. Debug mode is for development only.

### 2. How do you serve static files like images or CSS in Flask?

```python
from flask import Flask, render_template
app = Flask(__name__)

@app.get('/')
def home():
    return render_template('index.html')

# In templates/index.html:
# <link rel="stylesheet" href="{{ url_for('static', filename='style.css') }}">
```

Place CSS in static/style.css and HTML in templates/index.html.

### 3. How do you define different routes with different HTTP methods in Flask?

```python
from flask import Flask, request, jsonify
app = Flask(__name__)

@app.route('/items', methods=['GET', 'POST'])
def items():
    if request.method == 'GET':
        return jsonify(items=[])
    return jsonify(message='Item received'), 201
```

The same URL can handle GET and POST differently.

### 4. How do you render HTML templates in Flask?

```python
from flask import Flask, render_template
app = Flask(__name__)

@app.get('/')
def index():
    return render_template('index.html', name='Student')

# templates/index.html:
# <h1>Hello, {{ name }}</h1>
```

Flask uses Jinja templates from the templates/ directory.

### 5. How can you generate URLs for routes in Flask using url_for?

```python
from flask import Flask, url_for
app = Flask(__name__)

@app.get('/home')
def home():
    return 'Home'

@app.get('/link')
def link():
    return url_for('home')
```

The /link endpoint returns /home.

### 6. How do you handle forms in Flask?

```python
from flask import Flask, request
app = Flask(__name__)

@app.route('/contact', methods=['GET', 'POST'])
def contact():
    if request.method == 'POST':
        name = request.form.get('name', '').strip()
        return f'Hello, {name}'
    return '<form method="post"><input name="name"><button>Send</button></form>'
```

request.form reads fields submitted with a standard HTML form.

### 7. How can you validate form data in Flask?

```python
from flask import Flask, request, jsonify
app = Flask(__name__)

@app.post('/register')
def register():
    email = request.form.get('email', '').strip()
    if not email or '@' not in email:
        return jsonify(error='Valid email required'), 400
    return jsonify(message='Valid input')
```

This is basic illustrative validation; use a robust validator and CSRF protection for real forms.

### 8. How do you manage sessions in Flask?

```python
from flask import Flask, session
app = Flask(__name__)
app.config['SECRET_KEY'] = 'replace-with-env-secret'

@app.get('/login-demo')
def login_demo():
    session['username'] = 'student'
    return 'Session saved'

@app.get('/profile')
def profile():
    return session.get('username', 'Guest')
```

Use a strong secret from an environment variable in production; never store passwords in sessions.

### 9. How do you redirect to a different route in Flask?

```python
from flask import Flask, redirect, url_for
app = Flask(__name__)

@app.get('/')
def index():
    return redirect(url_for('dashboard'))

@app.get('/dashboard')
def dashboard():
    return 'Dashboard'
```

redirect() sends an HTTP redirect to the generated URL.

### 10. How do you handle errors in Flask (e.g., 404)?

```python
from flask import Flask, jsonify
app = Flask(__name__)

@app.errorhandler(404)
def not_found(error):
    return jsonify(error='Resource not found'), 404
```

The handler produces a JSON error response for missing routes.

### 11. How do you structure a Flask app using Blueprints?

```python
from flask import Flask, Blueprint
users = Blueprint('users', __name__, url_prefix='/users')

@users.get('/')
def list_users():
    return 'Users list'

app = Flask(__name__)
app.register_blueprint(users)
```

In larger projects, define the blueprint in a separate module and register it in the app factory.

### 12. How do you define a custom Jinja filter in Flask?

```python
from flask import Flask, render_template_string
app = Flask(__name__)

@app.template_filter('reverse_text')
def reverse_text(value):
    return str(value)[::-1]

@app.get('/')
def home():
    return render_template_string('{{ name|reverse_text }}', name='Flask')
```

The filter outputs 'ksalF'.

### 13. How can you redirect with query parameters in Flask?

```python
from flask import Flask, redirect, url_for, request
app = Flask(__name__)

@app.get('/start')
def start():
    return redirect(url_for('result', status='success'))

@app.get('/result')
def result():
    return request.args.get('status', 'unknown')
```

The redirect destination is /result?status=success.

### 14. How do you return JSON responses in Flask?

```python
from flask import Flask, jsonify
app = Flask(__name__)

@app.get('/api/student')
def student():
    return jsonify(id=1, name='Asha'), 200
```

jsonify() sets the JSON content type.

### 15. How do you capture URL parameters in Flask?

```python
from flask import Flask, jsonify
app = Flask(__name__)

@app.get('/users/<int:user_id>')
def user_detail(user_id):
    return jsonify(user_id=user_id)
```

For /users/42, Flask converts user_id to integer 42.
