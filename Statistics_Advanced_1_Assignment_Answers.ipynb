## 1. What is a random variable in probability theory?

A random variable is a numerical value assigned to each possible outcome of a random experiment. Its value depends on chance, but the probabilities of its possible values can be described mathematically. For example, when a coin is tossed three times, X may represent the number of heads, so X can take the values 0, 1, 2, or 3. Random variables make it possible to calculate probabilities, averages, and variations in uncertain situations.

## 2. What are the types of random variables?

The two main types are discrete and continuous random variables. A discrete random variable takes countable values, such as the number of customers visiting a store or the number of defective items in a batch. A continuous random variable can take any value within an interval, such as a person's height, temperature, or waiting time. Discrete variables are described by probability mass functions, whereas continuous variables are described by probability density functions.

## 3. Explain the difference between discrete and continuous distributions.

A discrete probability distribution assigns probabilities to individual countable outcomes. For example, the binomial distribution describes how many successes occur in a fixed number of independent trials. The probabilities of all possible outcomes add up to 1. A continuous distribution describes values along a continuous scale, such as weights or measurement errors. For continuous variables, probability is calculated over an interval using the area under a density curve; the probability of one exact value is zero. The normal distribution is a common continuous distribution.

## 4. What is a binomial distribution, and how is it used in probability?

A binomial distribution models the number of successes in a fixed number (n) of independent trials, each having only two outcomes: success or failure. The probability of success (p) must stay constant across trials. Its probability formula is P(X = k) = C(n, k) p^k (1 − p)^(n − k). For example, it can estimate the probability of getting exactly 6 heads in 10 fair coin tosses. It is useful in quality control, surveys, and yes/no experiments.

## 5. What is the standard normal distribution, and why is it important?

The standard normal distribution is a bell-shaped, symmetric continuous probability distribution with mean 0 and standard deviation 1. A normal variable X can be converted into a standard normal score using Z = (X − μ) / σ. A positive Z-score indicates a value above the mean, while a negative score indicates one below it. Standardization allows measurements from different scales to be compared and helps calculate probabilities, confidence intervals, and statistical test results.

## 6. What is the Central Limit Theorem (CLT), and why is it critical in statistics?

The Central Limit Theorem states that, under suitable conditions, the distribution of sample means approaches a normal distribution as sample size increases, even when the original population is not normally distributed. If observations are independent and have a finite variance, the sampling distribution has mean μ and standard error σ / √n. The CLT is important because it supports approximate confidence intervals and hypothesis tests for population means. A sample size of 30 is a common guideline, not a universal requirement.

## 7. What is the significance of confidence intervals in statistical analysis?

A confidence interval provides a range of plausible values for an unknown population parameter, such as the population mean. For example, a 95% confidence interval for average sales might be ₹240 to ₹260. The 95% refers to the long-run coverage rate of the method: across repeated samples, about 95% of similarly constructed intervals would contain the true parameter. Confidence intervals show both an estimate and its uncertainty; narrower intervals usually indicate greater precision.

## 8. What is the concept of expected value in a probability distribution?

Expected value is the probability-weighted average of all possible outcomes of a random variable. For a discrete variable, E(X) = Σ xP(X = x); for a continuous variable, it is calculated using an integral of x times its probability density. For example, if a fair six-sided die is rolled, its expected value is (1+2+3+4+5+6)/6 = 3.5. Although 3.5 cannot occur in one roll, it represents the long-run average over many rolls. Expected value is useful in forecasting, risk assessment, and decision-making.

## 9. Generate 1000 normal random numbers (mean = 50, SD = 5), calculate their mean and SD, and draw a histogram.

NumPy can simulate normally distributed data with a specified mean and standard deviation. We use a fixed random seed so the experiment is reproducible. The calculated sample mean and standard deviation should be close to 50 and 5, but will not be exactly equal because of random sampling. A histogram displays the distribution and should look approximately bell-shaped.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
data = rng.normal(loc=50, scale=5, size=1000)
print(f"Mean: {np.mean(data):.2f}")
print(f"Standard deviation: {np.std(data, ddof=1):.2f}")

plt.hist(data, bins=30, edgecolor="black")
plt.title("Normal Distribution (1000 Observations)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

Expected printed output (seed 42): Mean: 49.86; Standard deviation: 4.95. A histogram window is displayed.

## 10. Retail sales: apply the CLT and calculate mean sales and a 95% confidence interval.

The question provides 20 daily sales observations, even though the scenario mentions two years of sales data. We calculate the sample mean and sample standard deviation from the 20 supplied values. Assuming these observations are reasonably independent and representative, we can estimate uncertainty using the standard error s / √n. Because the population standard deviation is unknown and n = 20, a t-based 95% confidence interval is more appropriate than relying on a normal approximation alone. The CLT helps explain why sample means tend toward normality with sufficiently large samples, but it does not guarantee a good approximation for every sample of 20.

```python
import numpy as np
from scipy.stats import t

daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
n = len(daily_sales)
mean_sales = np.mean(daily_sales)
s = np.std(daily_sales, ddof=1)
se = s / np.sqrt(n)
margin = t.ppf(0.975, df=n - 1) * se
print(f"Mean daily sales: {mean_sales:.2f}")
print(f"95% CI: ({mean_sales - margin:.2f}, {mean_sales + margin:.2f})")
```

Output: Mean daily sales: 248.25; 95% CI: (240.17, 256.33). Interpretation: the interval estimates the population mean daily sales under the stated sampling assumptions; it does not describe the full two-year trend.
